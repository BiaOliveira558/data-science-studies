### Exercicio 1

<small>
Resolva os sistemas lineares a seguir. Utilize o método np.linalg.solve().
<small>

```python
import numpy as np
```

```python
A = np.array([[2,3],[1,-1]])
B = np.array([[6],[8]])
print(A)
print(B)
```

```python
solucao = np.linalg.solve(A,B)
print(solucao)
```

```python
A = np.array([[3,1],[1,2]])
B = np.array([[9],[8]])
```

```python
solucao = np.linalg.solve(A,B)
print(solucao)
```
