### Exercicio 1

<small>
Encontre a inversa das matrizes abaixo. Use o método np.linalg.inv().
<small>

```python
import numpy as np
```

```python
A = np.array([[1,2],[3,4]])
B = np.array([[-1,3],[5,4]])
```

```python
inv_a = np.linalg.inv(A)
inv_b = np.linalg.inv(B)
```

```python
print(inv_a)
print(inv_b)
```
