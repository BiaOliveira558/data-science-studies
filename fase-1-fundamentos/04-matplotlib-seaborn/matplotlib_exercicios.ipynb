1. Gráfico de Linha (plt.plot())
Uso: Ideal para mostrar tendências ao longo do tempo (séries temporais) ou relações contínuas.

Destaque: Conecta pontos de dados por segmentos de reta.

2. Gráfico de Dispersão (plt.scatter())
Uso: Ótimo para observar a correlação entre duas variáveis numéricas e identificar outliers (pontos fora da curva).

Destaque: Mostra cada observação como um ponto isolado.

3. Gráfico de Barras (plt.bar() ou plt.barh())
Uso: Comparar quantidades entre diferentes categorias.

Destaque: As barras podem ser verticais ou horizontais (barh).

4. Histograma (plt.hist())
Uso: Analisar a distribuição de uma variável numérica (frequência).

Destaque: Agrupa os dados em intervalos (bins) e mostra quantos valores caem em cada um.

5. Boxplot (plt.boxplot())
Uso: Visualizar a estatística descritiva dos dados: mediana, quartetos e valores discrepantes.

Destaque: Excelente para comparar a dispersão entre vários grupos.

6. Gráfico de Pizza ou Rosca (plt.pie())
Uso: Mostrar a proporção de partes em relação a um todo.

Destaque: Deve ser usado com cautela (apenas para poucas categorias) para não dificultar a leitura.

### Exercícios 15-20

```python
df = pd.read_csv('datasets/Iris.csv')
df.drop(columns = 'Id', inplace = True)
```

```python
df
```

```python
frequency = df['Species'].value_counts()
```

```python
frequency
```

```python
plt.bar(x=frequency.index,height=frequency)
plt.title('Frequency of the three species in the Iris Dataset')
plt.ylim(0,80)
plt.show()
```

```python
plt.barh(y=frequency.index,width=frequency)
plt.title('Frequency of the three species in the Iris Dataset')
plt.show()
```

```python
explode = (0,0.1,0.1)
labels = frequency.index
plt.pie(frequency, explode = explode, shadow = True, autopct = '%1.2f%%', labels = labels)
plt.title('Pie chart', fontsize = 15)
plt.legend(labels, loc = 'upper left')
plt.show()
```
