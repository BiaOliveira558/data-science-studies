iat/at

at: acessa um único valor para um dado par de rótulos de linha / coluna

Adicionando colunas com insert()

```python
nomes = ['Carla','Joao','Rafael','Ana','Laís']
cidades = ['Sao paulo','Recife','Salvador','Salvador','Rio de janeiro']
matematica = [6,7,8,9,10]
biologia = [5,3,3,4,10]
data = {
    'Nome': nomes,
    'Cidade': cidades,
    'Matematica': matematica,
    'Biologia': biologia
}

df = pd.DataFrame(data)
```

```python
df
```

```python
gender = ['F','M','M','F','F']
```

```python
df.insert(loc = 1,column='Genero_', value= gender)
```

```python
df_notas = df.columns[3:].to_numpy()
```

```python
df_notas
```

```python
df[df_notas].mean(axis=1)
```

```python
df['media'] = df[df_notas].mean(axis=1)
```

```python
df
```

Função apply

```python
def classifica_aluno(media):
    if media < 6:
        return 'Reprovado'
    else:
        return 'Aprovado'
```

```python
df['resultado_final']=df['media'].apply(classifica_aluno)
```

```python
df
```

Função Map

Utilizamos geralmente quando queremos tranformar variaveis categoricas em quantitativas

```python
df['target']=df['resultado_final'].map({'Reprovado':0,'Aprovado':1})
```

```python
df
```

```python
df['Gender_status']=df['Genero_'].map({'F': 1,'M':2})
```

```python
df
```
