### Exercícios 1 - 5

```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('datasets/Iris.csv')
df.drop(columns = 'Id', inplace = True)
```

```python
sns.countplot(df)
```

```python
sns.countplot(x = 'Species', data = df)
```

```python
sns.countplot(x = 'Species', data = df)
plt.title('Frequency of the three species in the Iris Dataset')
plt.ylabel('Frequency')
plt.ylim(0,80)
```

```python
sns.scatterplot(x = 'SepalLengthCm', 
                y = 'SepalWidthCm' , 
                data = df, 
                hue = 'Species', 
                alpha = 0.5)
plt.title('Relationship between Sepal Length and Sepal Width', fontsize = 12)
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.show()
```

```python
sns.scatterplot(x = 'PetalLengthCm', 
                y = 'PetalWidthCm' , 
                data = df, 
                hue = 'Species', 
                alpha = 0.5)
plt.title('Relationship between Sepal Length and Sepal Width', fontsize = 12)
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.show()
```

```python
fig, (ax1, ax2, ax3, ax4) = plt.subplots(4,1, figsize = (10, 10))
sns.histplot(x = 'SepalLengthCm', data = df, ax = ax1)
ax1.set(title = 'Sepal Length histogram', ylabel = 'Frequency')

sns.histplot(x = 'SepalWidthCm', data = df, ax = ax2)
ax2.set(title = 'Sepal Width histogram', ylabel = 'Frequency')

sns.histplot(x = 'PetalLengthCm', data = df, ax = ax3)
ax3.set(title = 'Petal Length histogram', ylabel = 'Frequency')

sns.histplot(x = 'PetalWidthCm', data = df, ax = ax4)
ax4.set(title = 'Petal Width histogram', ylabel = 'Frequency')

plt.tight_layout(pad = 2)
```

```python
plt.figure(figsize = (8, 5))
sns.boxplot(data = df)
plt.title('Box plot')
plt.show()
```

```python
corr = df.corr(numeric_only=True)
sns.heatmap(corr, cmap = 'Blues', annot = True)
```

```python
sns.pairplot(data=df)
plt.show()
```

### Exercício 16-20

```python
np.random.seed(20)
n = 2_000  #size
data = {
    'A': np.random.normal(loc = 100, scale = 1, size = n),
    'B': np.random.normal(loc = 100, scale = 2, size = n),
    'C': np.random.normal(loc = 100, scale = 5, size = n),
    'D': np.random.lognormal(mean = 2, sigma = 1.1, size = n),
    'E': np.random.lognormal(mean = 2, sigma = 1.2, size = n),
    'F': np.random.lognormal(mean = 2, sigma = 1.3, size = n),
    'G': np.random.randn(n),
    'H': np.random.binomial(n = 10, p = 0.8, size = n)
}
 
df = pd.DataFrame(data)
```

```python
df
```

```python
sns.kdeplot(x = 'A', data = df)
sns.kdeplot(x = 'B', data = df)
sns.kdeplot(x = 'C', data = df)
plt.title('Distributions')
plt.legend(['A', 'B', 'C'], loc = 'upper left')
plt.xlabel('Values')
plt.grid(linestyle = '--', alpha = 0.5)
plt.show()
```

```python
sns.kdeplot(x = 'D', data = df)
sns.kdeplot(x = 'E', data = df)
sns.kdeplot(x = 'F', data = df)
plt.title('Distributions')
plt.legend(['D', 'E', 'F'], loc = 'upper left')
plt.xlabel('Values')
plt.grid(linestyle = '--', alpha = 0.5)
plt.show()
```

```python
sns.jointplot(x = 'A', y = 'B', data = df);
```

```python
sns.jointplot(x = 'D', y = 'F', data = df);
```

```python
for column in data:
    sns.histplot(x = column, data = df, kde = True)
    plt.title(column+' - histogram')
    plt.show()
```

```python
sns.violinplot(y='G',data=df)
plt.title('G - violin plot');
```

```python
sns.violinplot(y = 'H', data = df)
plt.title('H - violin plot');
```
