```python
import numpy as np
```

<small>
Obtenha o determinante das matrizes a seguir. Utilize o método np.linalg.det().
<small>

```python
A = np.array([[5]])
B = np.array([[1,2],[3,4]])
C = np.array([[-3,-8],[-5,-2]])
D = np.eye(2)
```

```python
det_A = np.linalg.det(A).round()
print(det_A)
```

```python
det_B = np.linalg.det(B).round()
print(det_B)
```

```python
det_C = np.linalg.det(C).round()
print(det_C)
```

```python
det_D = np.linalg.det(D).round()
print(det_D)
```

### Exercicio 2

<small>
Obtenha o determinante das matrizes a seguir. Utilize o método np.linalg.det().
<small>

```python
A = np.array([[3,1,4],[10,7,5],[1,4,4]])
B = np.eye(3)
C = np.diag([5,10,2])
D = np.diag([3,7,11,2])
E = np.array([[5,0,13,12],[1,-1,25,7],[6,2,1,20],[11,4,0,-9]])
```

```python
det_A = np.linalg.det(A).round()
det_B = np.linalg.det(B).round()
det_C = np.linalg.det(C).round()
det_D = np.linalg.det(D).round()
det_E = np.linalg.det(E).round()

print(det_A)
print(det_B)
print(det_C)
print(det_D)
print(det_E)
```

### Exercicio 3

<small>
Dadas as matrizes A e B, calcule:

detA+detB

det(A+B)
<small>

```python
A = np.array([[1,2],[1,0]])
B = np.array([[3,-1],[0,1]])
det_A = np.linalg.det(A).round()
det_B = np.linalg.det(B).round()
print(det_A+det_B)
print(np.linalg.det(A+B).round())
```
